## Quick reference: arithmetic operators

| Operator | Purpose | Example |
|---|---|---|
| `+` | Addition | `8 + 2` |
| `-` | Subtraction | `8 - 2` |
| `*` | Multiplication | `8 * 2` |
| `/` | Division | `8 / 2` |
| `//` | Floor division | `17 // 5` |
| `%` | Remainder | `17 % 5` |
| `**` | Exponentiation | `2 ** 5` |

### Precedence from higher to lower

1. Parentheses: `()`
2. Exponentiation: `**`
3. Multiplication group: `*`, `/`, `//`, `%`
4. Addition group: `+`, `-`

When operators have the same precedence, Python generally evaluates from left to right. Exponentiation is an important exception because chained exponentiation binds from right to left.

````python
# A short precedence demonstration
without_parentheses = 2 + 3 * 4
with_parentheses = (2 + 3) * 4

print("Without parentheses:", without_parentheses)
print("With parentheses:", with_parentheses)
````

# Example 1: Satellite Storage Planning


### Problem
An Earth-observation satellite captures **275 images per day**. Each image occupies **48 MB**, and the mission operates for **365 days**.

Calculate:
1. the total number of images;
2. total storage in MB;
3. total storage in GB;
4. total storage in TB.

### Solution plan
1. Multiply images per day by mission days.
2. Multiply total images by the size of one image.
3. Convert MB to GB by dividing by 1024.
4. Convert GB to TB by dividing by 1024.

### Formula
- `total_images = images_per_day * days`
- `storage_mb = total_images * image_size_mb`
- `storage_gb = storage_mb / 1024`
- `storage_tb = storage_gb / 1024`

````python
images_per_day = 275
image_size_mb = 48
days = 365

total_images = images_per_day * days
storage_mb = total_images * image_size_mb
storage_gb = storage_mb / 1024
storage_tb = storage_gb / 1024

print("Total images: {total_images:,}")
print("Storage: {storage_mb:,} MB")
print("Storage: {storage_gb:,.2f} GB")
print("Storage: {storage_tb:,.2f} TB")
````

### Interpretation
The problem is solved as a calculation chain. Each intermediate variable answers one stage of the problem and becomes the input to the next stage. This is easier to inspect than placing the entire calculation in one long expression.

# Example 2: Mission Elapsed-Time Breakdown


### Problem
A spacecraft has been operating for **4,589 minutes**. Express the elapsed time as complete days, remaining hours, and remaining minutes.

### Solution plan
1. A day contains `24 × 60 = 1440` minutes.
2. Use floor division to obtain complete days.
3. Use modulus to find minutes not included in those days.
4. Repeat floor division and modulus to separate hours and minutes.

### Formula
- `days = total_minutes // minutes_per_day`
- `minutes_after_days = total_minutes % minutes_per_day`
- `hours = minutes_after_days // 60`
- `minutes = minutes_after_days % 60`

````python
total_minutes = 4589
minutes_per_hour = 60
hours_per_day = 24
minutes_per_day = minutes_per_hour * hours_per_day

days = total_minutes // minutes_per_day
minutes_after_days = total_minutes % minutes_per_day
hours = minutes_after_days // minutes_per_hour
minutes = minutes_after_days % minutes_per_hour

print("Complete days:", days)
print("Remaining hours:", hours)
print("Remaining minutes:", minutes)
print(f"Formatted duration: {days} days, {hours} hours, {minutes} minutes")
````

### Interpretation
Floor division counts complete groups, while modulus returns what remains. Together, `//` and `%` are especially useful for time conversion and resource allocation.

# Example 3: Satellite Memory Allocation


### Problem
A satellite has captured **18,347 images**. One storage drive can hold **5,000 images**.

Calculate:
1. the number of completely filled drives;
2. the number of images left after filling those drives;
3. the total number of drives required to store every image.

### Solution plan
- Use `//` to count completely filled drives.
- Use `%` to count remaining images.
- A partially filled drive is also required, so calculate the total using the ceiling-division expression `(items + capacity - 1) // capacity`.

````python
total_images = 18347
drive_capacity = 5000

full_drives = total_images // drive_capacity
remaining_images = total_images % drive_capacity
drives_required = (total_images + drive_capacity - 1) // drive_capacity

print("Completely filled drives:", full_drives)
print("Images on the final partial drive:", remaining_images)
print("Total drives required:", drives_required)
````

### Interpretation
`full_drives` and `drives_required` answer different questions. Three drives can be completely filled, but four drives are needed to store all images. This distinction is common in planning problems.

# Example 4: Ground-Station Downlink Planner


### Problem
A satellite must transmit **96 GB** of data at **40 MB/s**. Each ground-station pass lasts **12 minutes**.

Calculate:
1. total transmission time in seconds and minutes;
2. number of complete 12-minute passes;
3. transmission time needed in the final partial pass;
4. total number of passes required.

Assume `1 GB = 1024 MB`.

### Solution plan
Convert data to MB, divide by the transfer rate, then use floor division and modulus with the pass duration.

````python
data_gb = 96
download_rate_mb_s = 40
pass_duration_minutes = 12

data_mb = data_gb * 1024
total_seconds = data_mb / download_rate_mb_s
total_minutes = total_seconds / 60
pass_duration_seconds = pass_duration_minutes * 60

complete_passes = int(total_seconds // pass_duration_seconds)
final_pass_seconds = total_seconds % pass_duration_seconds
passes_required = int((total_seconds + pass_duration_seconds - 1) // pass_duration_seconds)

print(f"Data volume: {data_mb:,.0f} MB")
print(f"Transmission time: {total_seconds:,.1f} seconds")
print(f"Transmission time: {total_minutes:,.2f} minutes")
print("Complete passes:", complete_passes)
print(f"Time in final partial pass: {final_pass_seconds:.1f} seconds")
print("Total passes required:", passes_required)
````

### Interpretation
The total duration is longer than three complete passes but shorter than four. Operational planning therefore requires four passes, even though the last pass is only partially used.

# Example 5: Spacecraft Power and Energy Budget


### Problem
A spacecraft operates four subsystems:

- camera: **120 W** for **8 h/day**;
- communication system: **85 W** for **5 h/day**;
- navigation system: **60 W** for **24 h/day**;
- onboard computer: **45 W** for **24 h/day**.

The solar array generates **3,600 Wh/day**. Calculate:
1. energy used by each subsystem;
2. total daily energy demand;
3. remaining daily energy;
4. percentage of generated energy consumed.

### Solution plan
Energy in watt-hours is `power × operating time`. Add subsystem energy values, subtract from generation, and calculate a percentage.

````python
camera_energy = 120 * 8
communication_energy = 85 * 5
navigation_energy = 60 * 24
computer_energy = 45 * 24
solar_generation = 3600

total_energy = camera_energy + communication_energy + navigation_energy + computer_energy
remaining_energy = solar_generation - total_energy
consumption_percent = total_energy / solar_generation * 100

print("Camera energy:", camera_energy, "Wh")
print("Communication energy:", communication_energy, "Wh")
print("Navigation energy:", navigation_energy, "Wh")
print("Computer energy:", computer_energy, "Wh")
print("Total daily demand:", total_energy, "Wh")
print("Remaining energy:", remaining_energy, "Wh")
print(f"Energy consumed: {consumption_percent:.2f}%")
````

### Interpretation
A power rating alone does not show daily energy use. Operating duration matters. A lower-power system running continuously can consume more energy than a higher-power system used briefly.

# Example 6: Deep-Space Communication Delay


### Problem
A spacecraft is **300,000,000 km** from Earth. Radio signals travel at approximately **300,000 km/s**. Mission control needs **45 seconds** to prepare a reply after receiving the spacecraft signal.

Calculate:
1. one-way signal delay in seconds and minutes;
2. pure round-trip signal delay;
3. total time from spacecraft transmission to receipt of the prepared Earth reply.

### Solution plan
Divide distance by signal speed for one-way delay, double it for round-trip propagation, and add the response-preparation time.

````python
distance_km = 300_000_000
signal_speed_km_s = 300_000
response_preparation_seconds = 45

one_way_seconds = distance_km / signal_speed_km_s
one_way_minutes = one_way_seconds / 60
round_trip_seconds = one_way_seconds * 2
total_response_seconds = round_trip_seconds + response_preparation_seconds

print(f"One-way delay: {one_way_seconds:,.0f} seconds")
print(f"One-way delay: {one_way_minutes:.2f} minutes")
print(f"Round-trip propagation delay: {round_trip_seconds:,.0f} seconds")
print(f"Signal plus response preparation: {total_response_seconds:,.0f} seconds")
````

### Interpretation
Even when communication happens at the speed of light, large interplanetary distances create substantial delays. The calculation also demonstrates how a model can combine physical travel time and an operational delay.

# Example 7: Mission Budget with Contingency


### Problem
A mission has a base budget of **₹750 crore**. The base allocation is:

- launch: 40%;
- payload: 25%;
- ground segment: 20%;
- operations: 15%.

A **12% contingency reserve** is added on top of the base budget. Calculate each allocation, the contingency amount, and the revised total budget.

### Solution plan
Calculate percentage allocations from the base budget. Separately calculate contingency and add it to the base budget.

````python
base_budget_crore = 750

launch = base_budget_crore * 40 / 100
payload = base_budget_crore * 25 / 100
ground_segment = base_budget_crore * 20 / 100
operations = base_budget_crore * 15 / 100
contingency = base_budget_crore * 12 / 100
revised_budget = base_budget_crore + contingency
allocation_check = launch + payload + ground_segment + operations

print(f"Launch allocation: ₹{launch:.2f} crore")
print(f"Payload allocation: ₹{payload:.2f} crore")
print(f"Ground-segment allocation: ₹{ground_segment:.2f} crore")
print(f"Operations allocation: ₹{operations:.2f} crore")
print(f"Allocation check: ₹{allocation_check:.2f} crore")
print(f"Contingency reserve: ₹{contingency:.2f} crore")
print(f"Revised budget: ₹{revised_budget:.2f} crore")
````

### Interpretation
The allocation check should equal the base budget because the four percentages total 100%. Keeping the contingency separate avoids accidentally distributing more than the stated base budget.

# Example 8: High-Resolution Mapping Project


### Problem
A mapping project covers **15,000 km²**. One satellite image covers **36 km²**, but only **85%** of each image is usable because of overlap and edge effects. Processing one image takes **4.5 minutes**. Four computers process images in parallel at equal speed.

Calculate:
1. effective usable area per image;
2. number of images required;
3. total serial processing time;
4. approximate processing time with four computers.

### Solution plan
Adjust image coverage by the usable fraction. Use ceiling division to avoid requesting a fraction of an image. Then calculate processing workload and divide it across four computers.

````python
project_area_km2 = 15000
nominal_coverage_km2 = 36
usable_fraction = 0.85
processing_minutes_per_image = 4.5
computers = 4

effective_coverage_km2 = nominal_coverage_km2 * usable_fraction
images_required = int((project_area_km2 + effective_coverage_km2 - 1) // effective_coverage_km2)
serial_minutes = images_required * processing_minutes_per_image
parallel_minutes = serial_minutes / computers
parallel_hours = parallel_minutes / 60

print(f"Effective coverage per image: {effective_coverage_km2:.2f} km²")
print("Images required:", images_required)
print(f"Serial processing time: {serial_minutes:.1f} minutes")
print(f"Four-computer processing time: {parallel_minutes:.1f} minutes")
print(f"Four-computer processing time: {parallel_hours:.2f} hours")
````

### Interpretation
The usable fraction changes the planning result. Using nominal coverage would underestimate the number of images required. This example shows why formulas should represent real constraints, not only ideal values.

# Example 9: Earth-Observation Processing Cost


### Problem
A satellite collects **250 images per day** for **365 days**. Processing costs **₹12 per image**, storage costs **₹0.25 per image per month**, and the processed archive is retained for **18 months**. A 5% administrative charge is applied to the combined processing and storage cost.

Calculate:
1. total images;
2. total processing cost;
3. total storage cost;
4. subtotal;
5. administrative charge;
6. final project cost.

### Solution plan
Calculate the image count first because multiple later calculations depend on it. Storage cost depends on both image count and retention period.

````python
images_per_day = 250
days = 365
processing_cost_per_image = 12
storage_cost_per_image_month = 0.25
retention_months = 18
administrative_rate = 5 / 100

total_images = images_per_day * days
processing_cost = total_images * processing_cost_per_image
storage_cost = total_images * storage_cost_per_image_month * retention_months
subtotal = processing_cost + storage_cost
administrative_charge = subtotal * administrative_rate
final_cost = subtotal + administrative_charge

print(f"Total images: {total_images:,}")
print(f"Processing cost: ₹{processing_cost:,.2f}")
print(f"Storage cost: ₹{storage_cost:,.2f}")
print(f"Subtotal: ₹{subtotal:,.2f}")
print(f"Administrative charge: ₹{administrative_charge:,.2f}")
print(f"Final project cost: ₹{final_cost:,.2f}")
````

### Interpretation
This example demonstrates dependency between steps: the total image count is reused in processing and storage calculations, while the subtotal becomes the basis of the administrative charge.

# Example 10: Orbital Image-Scheduling Capacity


### Problem
A satellite completes **14 orbits per day** and can capture **37 images per orbit**. During each orbit, **5 images** are reserved for calibration and are not delivered to users. The mission operates for **180 days**.

Calculate:
1. gross images per day;
2. usable images per orbit;
3. usable images per day;
4. usable images over the mission;
5. percentage of captured images delivered to users.

### Solution plan
Calculate gross and usable production separately. The delivery percentage is usable production divided by gross production, multiplied by 100.

````python
orbits_per_day = 14
images_per_orbit = 37
calibration_images_per_orbit = 5
mission_days = 180

gross_images_per_day = orbits_per_day * images_per_orbit
usable_images_per_orbit = images_per_orbit - calibration_images_per_orbit
usable_images_per_day = orbits_per_day * usable_images_per_orbit
gross_mission_images = gross_images_per_day * mission_days
usable_mission_images = usable_images_per_day * mission_days
delivery_percent = usable_mission_images / gross_mission_images * 100

print("Gross images per day:", gross_images_per_day)
print("Usable images per orbit:", usable_images_per_orbit)
print("Usable images per day:", usable_images_per_day)
print(f"Usable mission images: {usable_mission_images:,}")
print(f"Delivered fraction: {delivery_percent:.2f}%")
````

### Interpretation
Not all acquired data becomes a user product. Calibration overhead reduces effective output, just as overlap or cloud cover can reduce usable remote-sensing observations.

# Example 11: Crew Logistics and Packaging


### Problem
A mission carries **37 crew members** for **180 days**. Each person requires **2.5 kg of supplies per day**. Supplies are packed in containers that hold **120 kg** each, and every container adds **8 kg** of packaging mass.

Calculate:
1. consumable supply mass;
2. number of containers required;
3. total packaging mass;
4. launch mass of supplies plus containers.

### Solution plan
Calculate consumable mass first. Use ceiling division for the number of containers because a partial container is still a complete physical container.

````python
crew = 37
mission_days = 180
supply_kg_person_day = 2.5
container_capacity_kg = 120
container_mass_kg = 8

consumable_mass_kg = crew * mission_days * supply_kg_person_day
containers_required = int((consumable_mass_kg + container_capacity_kg - 1) // container_capacity_kg)
packaging_mass_kg = containers_required * container_mass_kg
total_launch_mass_kg = consumable_mass_kg + packaging_mass_kg

print(f"Consumable mass: {consumable_mass_kg:,.1f} kg")
print("Containers required:", containers_required)
print(f"Packaging mass: {packaging_mass_kg:,.1f} kg")
print(f"Total launch mass: {total_launch_mass_kg:,.1f} kg")
````

### Interpretation
Operational calculations often include overhead. The packaging itself contributes launch mass, so the final requirement is larger than the crew's direct consumption.

# Example 12: Solar-Panel Output with Efficiency Losses


### Problem
A spacecraft has **18 solar panels**. Each panel has an area of **1.8 m²**. Solar irradiance is **1,361 W/m²**, panel efficiency is **29%**, and an additional **8% system loss** occurs in wiring and power conditioning.

Calculate:
1. total panel area;
2. incident solar power;
3. electrical power before system losses;
4. usable electrical power after losses.

### Solution plan
Multiply panel count by panel area, then apply efficiency and the remaining fraction after loss.

````python
panels = 18
area_per_panel_m2 = 1.8
irradiance_w_m2 = 1361
panel_efficiency = 29 / 100
system_loss = 8 / 100

total_area_m2 = panels * area_per_panel_m2
incident_power_w = total_area_m2 * irradiance_w_m2
electrical_power_before_loss_w = incident_power_w * panel_efficiency
usable_power_w = electrical_power_before_loss_w * (1 - system_loss)

print(f"Total panel area: {total_area_m2:.2f} m²")
print(f"Incident solar power: {incident_power_w:,.2f} W")
print(f"Electrical power before loss: {electrical_power_before_loss_w:,.2f} W")
print(f"Usable electrical power: {usable_power_w:,.2f} W")
````

### Interpretation
Percentages can be applied sequentially. A 29% conversion efficiency followed by an 8% system loss is not the same as simply subtracting 8 percentage points from 29%.

# Example 13: Rainfall Anomaly and Basin Volume


### Problem
A basin covers **2,400 km²**. Seasonal rainfall is **860 mm**, while the long-term normal is **740 mm**. Assume **35%** of the positive rainfall anomaly becomes effective surface water.

Calculate:
1. rainfall anomaly in mm;
2. anomaly in metres;
3. excess rainfall volume in cubic metres;
4. effective surface-water volume after applying the 35% coefficient.

Use `1 km² = 1,000,000 m²` and `1 mm = 0.001 m`.

### Solution plan
Find the rainfall difference, convert units, multiply depth by basin area, and then apply the effective-water fraction.

````python
basin_area_km2 = 2400
seasonal_rainfall_mm = 860
normal_rainfall_mm = 740
effective_fraction = 35 / 100

anomaly_mm = seasonal_rainfall_mm - normal_rainfall_mm
anomaly_m = anomaly_mm / 1000
basin_area_m2 = basin_area_km2 * 1_000_000
excess_volume_m3 = basin_area_m2 * anomaly_m
effective_volume_m3 = excess_volume_m3 * effective_fraction

print(f"Rainfall anomaly: {anomaly_mm:.1f} mm")
print(f"Rainfall anomaly: {anomaly_m:.3f} m")
print(f"Excess rainfall volume: {excess_volume_m3:,.0f} m³")
print(f"Effective surface-water volume: {effective_volume_m3:,.0f} m³")
````

### Interpretation
A modest rainfall-depth anomaly can correspond to a very large water volume when applied over a large area. Unit conversion is therefore a central part of the model, not a cosmetic step.

# Example 15: Integrated Mission Resource Dashboard


### Problem
A spacecraft will travel **540,000,000 km** at **90,000 km/h**. Fuel consumption is **650 kg/h**. The spacecraft carries **8 crew members**, each requiring **2.2 kg of supplies per day**. Communication signals travel at **300,000 km/s**.

Calculate:
1. travel time in hours and days;
2. complete travel days and remaining hours;
3. fuel required;
4. crew-supply requirement;
5. one-way communication delay at the stated distance;
6. total initial mission mass if the dry spacecraft mass is 24,000 kg.

### Solution plan
Use travel time as a shared intermediate result. It feeds both fuel and supply calculations. Use `//` and `%` to provide a human-readable time breakdown.

````python
distance_km = 540_000_000
speed_km_h = 90_000
fuel_rate_kg_h = 650
crew = 8
supply_kg_person_day = 2.2
signal_speed_km_s = 300_000
dry_spacecraft_mass_kg = 24_000

travel_hours = distance_km / speed_km_h
travel_days = travel_hours / 24
complete_days = int(travel_hours // 24)
remaining_hours = travel_hours % 24
fuel_required_kg = travel_hours * fuel_rate_kg_h
supply_required_kg = crew * supply_kg_person_day * travel_days
communication_delay_seconds = distance_km / signal_speed_km_s
initial_mass_kg = dry_spacecraft_mass_kg + fuel_required_kg + supply_required_kg

print(f"Travel time: {travel_hours:,.2f} hours")
print(f"Travel time: {travel_days:,.2f} days")
print(f"Time breakdown: {complete_days} complete days and {remaining_hours:.2f} hours")
print(f"Fuel required: {fuel_required_kg:,.2f} kg")
print(f"Crew supplies: {supply_required_kg:,.2f} kg")
print(f"One-way communication delay: {communication_delay_seconds:,.2f} seconds")
print(f"Initial mission mass: {initial_mass_kg:,.2f} kg")
````

### Interpretation
This capstone demonstrates dependency between variables. The travel time is not merely an output: it becomes an input to fuel and supply calculations. Intermediate variables make that dependency visible.

## Main ideas

- An **expression** combines values, variables, and operators to produce a result.
