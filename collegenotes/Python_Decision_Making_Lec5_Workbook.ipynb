## elif : Multiple Choices

````python
signal = "yellow"
if signal == "red":
    print("Stop")
elif signal == "yellow":
    print("Wait")
else:
    print("Go")
````

## Logical Operators

### AND

````python
age = 20
has_id = True
if age >= 18 and has_id:
    print("Entry Allowed")
````

### OR

````python
is_student = False
is_staff = True
if is_student or is_staff:
    print("Access Granted")
````

### NOT

````python
raining = False
if not raining:
    print("Go for a walk")
````

## Nested if Statements
A decision inside another decision.

````python
age = 20
has_id = True
if age >= 18:
    if has_id:
        print("Entry Allowed")
````
