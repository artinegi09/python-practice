```python
# Write e aprogmram to find the sum of following series
# 1 + 2 + 3 + 4 + 5 + ....... + num
# Rubber Duck Technique

num = input()
num = int(num)

counter = 1
ans = 0
while counter <= num:    
    ans = ans + counter
    print(f"Counter: {counter} , Ans: {ans}")
    counter = counter + 1
```
